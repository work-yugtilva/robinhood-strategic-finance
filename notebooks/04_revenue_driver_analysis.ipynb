# Revenue drivers and exact growth bridges
Activity per funded customer is an aggregate proxy, not product-user activity. Missing and acquisition-scope effects remain in the bridge residual.


```python
import csv
from pathlib import Path
root = Path.cwd()
mix = list(csv.DictReader((root / 'data/analysis/revenue_mix_quarterly.csv').open()))
bridges = list(csv.DictReader((root / 'data/analysis/revenue_growth_bridges.csv').open()))
[r for r in bridges if r['product'] == 'TOTAL_TRANSACTION']
```

## Check
For each total row, customer + activity + monetization + residual equals the change in transaction revenue. Formula and source limitations are documented in `reports/HISTORICAL_ANALYSIS.md`.


```python
for r in bridges:
    if r['product'] == 'TOTAL_TRANSACTION':
        effects = sum(float(r[k]) for k in ('customer_scale_effect_m','activity_effect_m','monetization_effect_m','mix_residual_effect_m'))
        assert abs(effects - float(r['change_m'])) < 1e-8
```

# Operating leverage and net interest context
Incremental margins are descriptive. Acquisition periods and small revenue changes require the interpretation flags in the table. The federal funds rate is external context, not Robinhood's realized yield.


```python
import csv
from pathlib import Path
root = Path.cwd()
leverage = list(csv.DictReader((root / 'data/analysis/operating_leverage_quarterly.csv').open()))
financial = list(csv.DictReader((root / 'data/analysis/financial_kpis_quarterly.csv').open()))
[(r['period'], r['incremental_operating_margin_yoy_pct'], r['interpretation_flag']) for r in leverage]
```

```python
[(r['period'], r['net_interest_revenue_m'], r['interest_asset_avg_balance_m'], r['interest_asset_implied_yield_pct'], r['fed_funds_quarter_average_pct']) for r in financial if r['interest_asset_avg_balance_m']]
```

## Interpretation
The matched interest-asset balance and revenue series is available for selected quarters only. Do not apply its implied yield to all net interest revenue. See `reports/HISTORICAL_ANALYSIS.md`.
